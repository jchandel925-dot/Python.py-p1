```python
def zipper_file(filename, strings):
    with open(filename, "w") as f:
        for i in range(0, len(strings), 2):
            a, b = strings[i], strings[i + 1]
            zipped = ""
            for j in range(max(len(a), len(b))):
                if j < len(a):
                    zipped += a[j]
                if j < len(b):
                    zipped += b[j]
            f.write(zipped + "\n")
```

```python
def reformat_date(dates):
    months = ["January", "February", "March", "April", "May", "June",
              "July", "August", "September", "October", "November", "December"]
    result = []
    for date in dates:
        mm, dd, yyyy = date.split("/")
        result.append(f"Office Hours on {months[int(mm) - 1]} {dd}, {yyyy}")
    return result
```

```python
if __name__ == "__main__":
    zipper_file("out.txt", ["abc", "123", "hi", "yo"])
    print(reformat_date(["01/07/2024", "02/15/2025"]))
```
